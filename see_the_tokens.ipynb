# What a real tokenizer does to your text

The tokenizer of a small open language model (Qwen2.5-1.5B-Instruct). It only downloads the tokenizer, not the model.

```python
!pip -q install tokenizers
```

```python
from tokenizers import Tokenizer

tok = Tokenizer.from_pretrained("Qwen/Qwen2.5-1.5B-Instruct")

def show(text):
    ids = tok.encode(text, add_special_tokens=False).ids
    pieces = [tok.decode([i]) for i in ids]
    print(f"{len(ids):3d} tokens  {pieces}  {ids}")

print(tok.get_vocab_size(with_added_tokens=True), "different tokens")
```

## One word, one number

Inside a sentence a word carries the space in front of it.

```python
show(" strawberry")
show(" strawbery")
show("s t r a w b e r r y")
show("How many r's are in the word strawberry?")
```

## The same request in two languages

```python
show("Can I have a cup of coffee with milk, please?")
show("هل يمكنني الحصول على فنجان قهوة مع حليب، من فضلك؟")
```
